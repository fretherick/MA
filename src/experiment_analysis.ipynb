```python
exp = pd.read_csv(r"C:\Users\frede\OneDrive\Dokumente\GitHub\SIGSOM_Paper\table3.csv", encoding='utf-16')
exp2 = pd.read_csv(r"C:\Users\frede\OneDrive\Dokumente\GitHub\SIGSOM_Paper\table3_extra.csv", encoding='utf-16')

exp = pd.concat([exp, exp2], ignore_index=True)
```

```python
exp = exp[exp['State'] == 'Success']
exp[["simulation_experiments.hmm.params.len_path", "overall_accuracy", "overall_std", "simulation_experiments.clustering.model",'simulation_experiments.hmm.params.dt']]
```

```python
# Assume you’ve already done:
# exp = exp[exp['State'] == 'Success']

# 1. Pick just the four columns you care about:
cols = [
    "simulation_experiments.hmm.params.len_path",
    "overall_accuracy",
    "overall_std",
    "simulation_experiments.clustering.model",
    "simulation_experiments.hmm.params.dt"
]
df = exp[cols]

# 2. Find the unique dt-values:
dts = df["simulation_experiments.hmm.params.dt"].unique()

# 3. For each dt, make a sorted table:
tables = {}
for dt in sorted(dts):
    sub = df[df["simulation_experiments.hmm.params.dt"] == dt] \
            .drop(columns="simulation_experiments.hmm.params.dt") \
            .sort_values(by="simulation_experiments.hmm.params.len_path", ascending=True) \
            .reset_index(drop=True)
    tables[dt] = sub

# Now `tables` is a dict: e.g. tables[0.01], tables[0.1], tables[1.0], etc.
# Each one is already sorted by len_path ascending.

# If you want to see them in Jupyter, you could do:
for dt, table in tables.items():
    print(f"--- dt = {dt} ---")
    display(table)
```

```python
# Filter only for 'jump' or 'hmm' models
df = df[df["simulation_experiments.clustering.model"].isin(["jump", "hmm"])]

# Get unique dt-values
dts = df["simulation_experiments.hmm.params.dt"].unique()

# Create sorted tables for each dt
tables = {}
for dt in sorted(dts):
    sub = df[df["simulation_experiments.hmm.params.dt"] == dt] \
            .drop(columns="simulation_experiments.hmm.params.dt") \
            .sort_values(by="simulation_experiments.hmm.params.len_path", ascending=True) \
            .reset_index(drop=True)
    tables[dt] = sub

# Optionally display the result in a notebook
for dt, table in tables.items():
    print(f"--- dt = {dt} ---")
    display(table)
```
